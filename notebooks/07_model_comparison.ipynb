## Why Does the CNN Perform Better?

The CNN has access to significantly more information than the Random Forest.

Instead of using only 13 averaged MFCC values, it processes the complete Mel-spectrogram.

This allows the network to learn:

- harmonic structures
- frequency patterns
- rhythmic information
- local time-frequency relationships

As a result, the CNN can capture musical characteristics that are lost when averaging MFCC features.

## Advantages and Disadvantages

## Lessons Learned

Several important insights emerged during the project:

- Data representation strongly influences model performance.
- MFCC features provide a useful baseline but discard information.
- CNNs can learn useful patterns directly from Mel-spectrograms.
- More expressive models generally require more computation.
- Model evaluation should go beyond accuracy alone.
